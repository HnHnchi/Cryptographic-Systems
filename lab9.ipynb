```python
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad, unpad
import base64

# Input data
plaintext = "Cryptography section 3"
key = "AES-128"  # This key is too short, must be 16 bytes

# Pad the key to 16 bytes (128 bits)
key_16 = key.ljust(16, ' ')  # pad with spaces to 16 bytes
key_bytes = key_16.encode('utf-8')

# Create AES cipher in ECB mode
cipher = AES.new(key_bytes, AES.MODE_ECB)

# Pad plaintext to 16-byte blocks
plaintext_bytes = plaintext.encode('utf-8')
padded_plaintext = pad(plaintext_bytes, AES.block_size)

# Encrypt
ciphertext = cipher.encrypt(padded_plaintext)
ciphertext_b64 = base64.b64encode(ciphertext)  # for readable format

print("Encrypted (Base64):", ciphertext_b64.decode())

# Decrypt
decrypted_padded = cipher.decrypt(ciphertext)
decrypted = unpad(decrypted_padded, AES.block_size)

print("Decrypted:", decrypted.decode())
```

```python
from Crypto.Cipher import DES
from Crypto.Util.Padding import pad, unpad
import base64

# Shared secret key (must be 8 bytes for DES)
shared_key = b'8bytekey'

# Function to encrypt a message
def encrypt_message(message, key):
    cipher = DES.new(key, DES.MODE_ECB)
    padded_text = pad(message.encode(), DES.block_size)
    encrypted = cipher.encrypt(padded_text)
    return base64.b64encode(encrypted).decode()

# Function to decrypt a message
def decrypt_message(encrypted_b64, key):
    cipher = DES.new(key, DES.MODE_ECB)
    encrypted = base64.b64decode(encrypted_b64)
    decrypted = unpad(cipher.decrypt(encrypted), DES.block_size)
    return decrypted.decode()

# Simulate user input
def simulate_chat():
    while True:
        sender = input("\nWho is sending the message? (User1/User2 or 'exit' to quit): ")
        if sender.lower() == 'exit':
            break
        message = input(f"{sender}'s message: ")

        # Encrypt the message
        encrypted_msg = encrypt_message(message, shared_key)
        print(f"Encrypted message: {encrypted_msg}")

        # Decrypt the message at receiver's end
        receiver = 'User2' if sender == 'User1' else 'User1'
        decrypted_msg = decrypt_message(encrypted_msg, shared_key)
        print(f"{receiver} received and decrypted: {decrypted_msg}")

# Start the chat simulation
simulate_chat()
```

```python
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad, unpad
import base64

# Message and fixed key
message = "Hello World!"
key = b'1234567890abcdef'  # 16-byte key for AES-128

# AES encryption in ECB mode
cipher = AES.new(key, AES.MODE_ECB)

# Pad message to block size
padded_msg = pad(message.encode(), AES.block_size)

# Encrypt and encode in base64 for readability
encrypted = cipher.encrypt(padded_msg)
encrypted_b64 = base64.b64encode(encrypted).decode()

print("🔐 Encrypted (Base64):", encrypted_b64)

# Decrypt
cipher_dec = AES.new(key, AES.MODE_ECB)
decrypted = unpad(cipher_dec.decrypt(base64.b64decode(encrypted_b64)), AES.block_size)

print("🔓 Decrypted:", decrypted.decode())
```
